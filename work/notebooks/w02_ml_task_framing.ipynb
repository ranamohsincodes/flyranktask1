## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

A fixed rule (like "stale AND visible") only asks one or two yes/no questions with hand-picked thresholds. It can work, but it can't adapt when the real pattern is more nuanced — for example, when the boundary between "needs review" and "fine" depends on a combination of signals (position, content age, impressions) in ways a human wouldn't easily guess. In Week 2, the decision tree showed it could find thresholds like impressions_90d <= 5.50 that weren't obvious from intuition alone. ML can also be retrained as the data shifts, while a hand rule stays fixed unless someone manually updates it.
